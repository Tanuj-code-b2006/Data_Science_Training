```python
#WAP to print Fibonacci series

def fibonacci(n):
    a= 0
    b= 1
    series = []
    for i in range(n):
        series.append(a)
        a= b
        b=a + b
    return series

num_terms = int(input("Enter number of terms: "))
print("Fibonacci series:", fibonacci(num_terms))
```

```python
# WAP Program to calculate gross income of a family
n = int(input("Enter number of family members: "))
gross_income = 0

for i in range(n):
    income = float(input(f"Enter income of member {i+1}: "))
    gross_income += income                                                                   

print("Gross Income of the family =", gross_income)
```

```python
#WAP to implement Simple Calculator

while True:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    op = input("Enter operation (+, -, *, /): ")

    if op == '+':
        print("Result =", num1 + num2)
    elif op == '-':
        print("Result =", num1 - num2)
    elif op == '*':
        print("Result =", num1 * num2)
    elif op == '/':
        if num2 != 0:
            print("Result =", num1 / num2)
        else:
            print("Error! Division by zero.")
    else:
        print("Invalid operation")
        
    choice = input("Do you want to continue? (y/n): ").lower()
    if choice != 'y':
        print("Calculator exited.")
        break
```
