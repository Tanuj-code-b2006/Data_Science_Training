```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open(r"C:\Users\bhard\Downloads\hihihi.jpg")
img_arr = np.array(img)


img_1 = np.array(img)
img_1[:,:,0] = np.clip(img_arr[:,:,0] + 0, 0, 255)


img_2 = np.array(img)
img_2[:,255,:] = np.clip(img_arr[:,255,:] + 0, 0, 255)


img_3 = np.array(img)
img_3[0,:,:] = np.clip(img_arr[0,:,:] + 90, 0, 255)


img_4 = np.array(img)
img_4[:,0,:120] = np.clip(img_arr[:,0,:120] + 0, 0, 255)

# Subplots
fig, axes = plt.subplots(2, 2, figsize=(10,10))

axes[0,0].imshow(img_1); axes[0,0].set_title("Image 1"); axes[0,0].axis("off")
axes[0,1].imshow(img_2); axes[0,1].set_title("Image 2"); axes[0,1].axis("off")
axes[1,0].imshow(img_3); axes[1,0].set_title("Image 3"); axes[1,0].axis("off")
axes[1,1].imshow(img_4); axes[1,1].set_title("Image 4"); axes[1,1].axis("off")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open(r"C:\Users\bhard\Downloads\hihihi.jpg")
img_arr = np.array(img)

# 1. Remove Red
img_1 = np.array(img_arr)
img_1[:,:,0] = 0  

# 2. Remove Green
img_2 = np.array(img_arr)
img_2[:,:,1] = 0  

# 3. Remove Blue
img_3 = np.array(img_arr)
img_3[:,:,2] = 0

img_4 = np.array(img_arr)
img_4[:,:,0] = 0   
img_4[:,:,1] = 0   


fig, axes = plt.subplots(2, 2, figsize=(10,10))

axes[0,0].imshow(img_1); axes[0,0].set_title("No Red"); axes[0,0].axis("off")
axes[0,1].imshow(img_2); axes[0,1].set_title("No Green"); axes[0,1].axis("off")
axes[1,0].imshow(img_3); axes[1,0].set_title("No Blue"); axes[1,0].axis("off")
axes[1,1].imshow(img_4); axes[1,1].set_title("No Yellow"); axes[1,1].axis("off")

plt.tight_layout()
plt.show()
```
